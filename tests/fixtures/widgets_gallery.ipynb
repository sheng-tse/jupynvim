```python
# W:image
import struct as _st, zlib as _zl
def _chunk(t, d): return _st.pack('>I', len(d)) + t + d + _st.pack('>I', _zl.crc32(t + d))
_rows = b''.join(b'\x00' + b''.join(bytes((x * 255 // 160, 120, 255 - x * 255 // 160)) for x in range(160)) for _ in range(80))
_png = (b'\x89PNG\r\n\x1a\n' + _chunk(b'IHDR', _st.pack('>IIBBBBB', 160, 80, 8, 2, 0, 0, 0))
        + _chunk(b'IDAT', _zl.compress(_rows)) + _chunk(b'IEND', b''))
w.Image(value=_png, format='png', width=160, height=80)
```

```python
# W:applayout
def _b(d, s):
    return w.Button(description=d, button_style=s, layout=w.Layout(width='auto', height='auto'))
w.AppLayout(header=_b('Header', 'success'), left_sidebar=_b('Left', 'info'), center=_b('Center', 'warning'),
            right_sidebar=_b('Right', 'info'), footer=_b('Footer', 'success'), height='220px')
```

```python
# W:gridspeclayout
_g = w.GridspecLayout(3, 3, height='180px')
_g[0, :] = w.Button(description='top row', button_style='success', layout=w.Layout(width='auto', height='auto'))
_g[1:, 0] = w.Button(description='left col', button_style='info', layout=w.Layout(width='auto', height='auto'))
_g[1:, 1:] = w.Button(description='big', button_style='warning', layout=w.Layout(width='auto', height='auto'))
_g
```

```python
# W:output
import matplotlib
import matplotlib.pyplot as plt
_out = w.Output(layout={'border': '1px solid black'})
with _out:
    print('printed inside an Output widget')
    _fig, _ax = plt.subplots(figsize=(4, 2.5))
    _ax.plot([0, 1, 2, 3], [1, 3, 2, 4], marker='o')
    _ax.set_title('Output widget plot')
    plt.show()
_out
```

```python
# W:interact
from ipywidgets import interact
@interact(x=(0, 10), color=['red', 'green', 'blue'])
def _f(x=5, color='green'):
    print(f'x={x} squared={x * x} color={color}')
```

```python
# W:tqdm
from tqdm.notebook import tqdm as _tq
from tqdm.auto import tqdm as _tqa
for _i in _tq(range(40), desc='tqdm.notebook done'):
    time.sleep(0.01)
_t = _tqa(total=100, desc='tqdm.auto partial')
_t.update(60)
_ = _t.refresh()
```
